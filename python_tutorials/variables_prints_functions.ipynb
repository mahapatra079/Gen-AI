```python
# Variable Naming Convention:- 

# Use descriptive names
student_name = "Amit"
student_age = 20

# Use lowercase letters and underscores (snake_case)
favorite_subject = "Python"
total_marks = 450

# Names can contain numbers, but cannot start with a number
score1 = 95       # Valid
# 1score = 95     # Invalid

# Names are case-sensitive
age = 20
Age = 25          # Different variable from age

# Use uppercase names for constants by convention
MAX_SCORE = 100
PI = 3.14159

# Do not use Python keywords as variable names
# class = "Python"  # Invalid: class is a keyword

# Avoid spaces and special characters
first_name = "Amit"  # Valid
# first name = "Amit"  # Invalid: spaces aren't allowed
# first-name = "Amit"  # Invalid: hyphens aren't allowed
```

```python
# Input() can take the input from the user side and then can store that input into a variable.
# Input function only takes the input in a string format until and unless we are not going to explicitly take the input in  integer or other format 
```

```python
## building a simple calculator

variable1 = input("Enter first number: ")
variable2 = input("Enter second number: ")

print("the addition of two numbers is: ", int(variable1) + int(variable2))
print("the subtraction of two numbers is: ", int(variable1) - int(variable2))
print("the multiplication of two numbers is: ", int(variable1) * int(variable2))
print("the division of two numbers is: ", int(variable1) / int(variable2))
```
