# Data Types

## Data types are classification of the data that tells the compiler or the interpreter that what kind of values we can store into a variable and how much amount of memory will be consumed to store the data and the operations it can perform.

```python
age = input("enter the age:")
print("The age entered is:", age)
print(type(age))
```

# Data types in python

# Example of data types in python are as follows:
    1. int 
    2. float
    3. complex
    4. str
    5. bool
    6. list
    7. tuple
    8. set
    9. dict

```python
x = 5
print(type(x))
```

```python
x = "Hello World"
print(type(x))
```

```python
x = 20.5
print(type(x))
```

```python
x = True
print(type(x))
```

```python
x = {"name" : "John", "age" : 36}
print(x)
print(type(x))
```

```python
x = list(("apple", "banana", "cherry"))
print(x)
print(type(x))
```
