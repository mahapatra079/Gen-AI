# File Handling

# File handling is an important part of any web application.

# Python has several functions for creating, reading, updating, and deleting files.

# Types of Files - 

1. Text File - .txt, .docx,etc
2. Binary File - .png, .mp4, .mp3, .mov

# The key function for working with files in Python is the open() function.

# The open() function takes two parameters; filename, and mode.

# There are four different methods (modes) for opening a file:

  1. "r" - Read - Default value. Opens a file for reading, error if the file does not exist

  2. "a" - Append - Opens a file for appending, creates the file if it does not exist

  3. "w" - Write - Opens a file for writing, creates the file if it does not exist

  4. "x" - Create - Creates the specified file, returns an error if the file exists


# In addition you can specify if the file should be handled as binary or text mode

   1. "t" - Text - Default value. Text mode
   2. "b" - Binary - Binary mode (e.g. images)

# Syntax - To open a file for reading it is enough to specify the name of the file:

- f = open("demofile.txt")

#  The code above is the same as:

- f = open("demofile.txt", "rt")

# Because "r" for read, and "t" for text are the default values, you do not need to specify them.

# Note: Make sure the file exists, or else you will get an error.

```python
# example

f = open("demofile.txt", "rt")
content = f.read()
print(content)
f.close()
```

```python
# example - write() ---> helps us out in writing the content into the file 

f = open('example.txt','w')
f.write('Hello we are now writng the content into our file')
f.close()
```

```python
## Appending the content to our file 

f = open('example.txt','a')
f.write('\n We are learninga about GEN AI pre requisite python')
f.close()
```

#  with statement

# with - It's used for opening the file as object and there is no requirement for closing the file as well.

# with blocks close the file automatically

```python
# example  - with

with open('example.txt','r') as f:
    content_data=f.read()
    print(content_data)
```

```python
# example

# Create the file and write some text
with open("example.txt", "w", encoding="utf-8") as file:
    file.write("Hello! This text is in example.txt.")

# Open the file and read its contents
with open("example.txt", "r", encoding="utf-8") as file:
    content = file.read()

print(content)
```
