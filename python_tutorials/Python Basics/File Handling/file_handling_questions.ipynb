```python
## example - Read the content from the example.txt file and then writing the content into another file 

with open('source.txt','r') as source:
    content = source.read()

with open('destination.txt','w') as destination:
    destination.write(content)
```

# readline() reads one line at a time from a file, including the trailing \n. Each call moves the file position forward to the next line.

```python
## Reading the total lines, Total words , Total characters from a file -->readlines()

def count_lines(file):
    with open(file,'r') as files:
        lines=files.readline()
        total_lines=len(lines)
        total_words=sum(len(line.split()) for line in lines)
        total_characters = sum(len(line) for line in lines)
    return total_lines,total_words, total_characters
file_path='total_data.txt'
Total_lines,Total_words, Total_characters= count_lines(file_path)
print(f"Total lines are: {Total_lines}, Total words are : {Total_words} and Total Characters are: {Total_characters}")
```

# W+ --> First of all a W is their its states that first writing operations will be done and then Reading operation will be done

# seek() --> it is going to help us out reading the content after writing from starting only

# seek() moves the file position (the cursor) to a chosen location, so the next read or write starts from there. Its companion tell() reports the current position.

```python
## Write the content into a file and then after writing the content into that file we are going to read the content from that file as well -->Seek()

with open("sample.txt", "w+", encoding="utf-8") as write_read_file:
    write_read_file.write("\n we are looking  into w+ operations")
    write_read_file.write('\n it means writing and then reading')
    write_read_file.seek(0)
    content_coomplete_data = write_read_file.read()
    print(content_coomplete_data)
```

## open(..., "w+"): opens for reading and writing. - If the file exists, its contents are erased immediately. If it doesn’t exist, it is created.

## write(...) (twice): both strings are added one after another. The file position is now at the end.

## seek(0): moves the position back to the start. This is the key step.

## read(): reads from the start to the end.

## print(...): shows the result.

## Note - After writing, the position sits at the end of the file. Without seek(0), read() would return an empty string
