# Loop Statement

# Loops let you run a block of code repeatedly. Python has two: for and while.

# Loop statements are used for iterating over a series of numbers or through iterables.


```python
# example - print numbers from 1 to 5

print(1)
print(2)
print(3)
print(4)
print(5)
```

# range()

# range() -  range function can be used for working as our iterable

# Generates a sequence of numbers.

# range() generates a sequence of numbers on demand. It has three forms:

# Syntax  - range(stop)                 # 0 up to stop-1
#         - range(start, stop)          # start up to stop-1
#         - range(start, stop, step)    # start up to stop-1, jumping by step



```python
# range() example 1 - 

for i in range(5):
    print(i)   
```

```python
# range() example 2

for i in range(1, 6):
    print(i)    
```

```python
# range example 3

for i in range(0, 10, 2):
    print(i) 
```

```python
# range - Counting backwards

for i in range(5, 0, -1):
    print(i) 
```

```python
for  i in range (0,100,5):
    print(i)
```

```python
# Looping by index

fruits = ["apple", "banana", "cherry"]

for i in range(len(fruits)):
    print(i, fruits[i])
```

# Looping statements, we can iterate through iterable like String as well

```python
# example for Stirng 

Name = "Amit Kumar Mahapatra"
for i in Name:
    print(i)
```

```python
# Given a list of orders amount: orders = [1200, 2500, 800, 1750, 3000] use a for loop
# to apply the discount rules from Task1 to each order and print summary table 
# showing order: order_amount -> discount% -> final_amount.

orders = [1200, 2500, 800, 1750, 3000]

print("Order amount -> Discount % -> Final amount")

for order_amount in orders:
    if order_amount >= 2000:
        discount_rate = 0.15
    elif order_amount >= 1500:
        discount_rate = 0.10
    elif order_amount >= 1000:
        discount_rate = 0.07
    else:
        discount_rate = 0

    final_amount = order_amount * (1 - discount_rate)
    print(f"{order_amount} -> {discount_rate * 100:.0f}% -> {final_amount:.2f}")
```

```python
# Also compute and print total revenue after discount.
  # Extra(optional): Print the numbers of orders that recieved a discount (discount > 0)


orders = [1200, 2500, 800, 1750, 3000]
total_revenue = 0
discounted_orders = 0

print("Order amount -> Discount % -> Final amount")

for order_amount in orders:
    if order_amount >= 2000:
        discount_rate = 0.15
    elif order_amount >= 1500:
        discount_rate = 0.10
    elif order_amount >= 1000:
        discount_rate = 0.07
    else:
        discount_rate = 0

    final_amount = order_amount * (1 - discount_rate)
    total_revenue += final_amount

    if discount_rate > 0:
        discounted_orders += 1

    print(f"{order_amount} -> {discount_rate * 100:.0f}% -> {final_amount:.2f}")

print(f"\nTotal revenue after discount: {total_revenue:.2f}")
print(f"Number of discounted orders: {discounted_orders}")
```
