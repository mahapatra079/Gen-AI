# OS Library

# OS Library is able to interact with our operating system and can let us work with multiple directories and files.

# The os module is part of Python’s standard library (no installation needed). It lets your code interact with the operating system: files, directories, environment variables, and processes.

# import os

# Working with directories
1. os.getcwd()                          # current working directory
2. os.chdir("/path/to/dir")             # change directory
3. os.listdir(".")                      # list files and folders
4. os.mkdir("new_folder")               # create a single directory
5. os.makedirs("a/b/c", exist_ok=True)  # create nested directories
6. os.rmdir("empty_folder")             # remove an empty directory
7. os.removedirs("a/b/c")               # remove nested empty directories

```python
# example - os.getcwd() 

import os

current_path = os.getcwd()
print("the current working directory is : ", current_path)
```

```python
## mkdir - Creates a directory (with a specified mode)
new_path="New_Folder"
os.mkdir(new_path)
print(f"the new directory {new_path} is created successfully")
```

```python
## listdir() - listing all of the directories currently

All_items = os.listdir('.')
print(All_items)
```

```python
# Joining 2 paths together 

new_path=os.path.join("File Handling", "New_folder")
os.mkdir(new_path)
print(f"the new directory {new_path} is created successfully")
```

```python
# example - os.path.join

dir_name = "Folder_Data"
file_name = "File_Data"
Full_path = os.path.join(dir_name,file_name)
print(f"The full path is :", {Full_path})
```

```python
# example - I want to join Multiple Path with the current path

current_complete_path = os.path.join(os.getcwd(),dir_name,file_name)
print(f"The complete path is : {current_complete_path}")
```

```python
## os.path.exits()

given_path="f:\Gen Ai\python_tutorials\Folder_Data\File_Data"

if os.path.exists(given_path):
    print("Yes the path exists")
else:
    print("This path does not exists")
```

```python
## os.path.isfile() - it checks that whether if the current path is a file or not 
## os.path.isdir() - it checks that whether if the current path is directory or not

path = "Total_data.txt"

if os.path.isfile(path):
    print("The current path belongs to a file")
elif os.path.isdir(path):
    print("The current path belongs to a directory")
else: 
    print("The current path is neighther a file or a folder")
```

```python
# example 

path = "New_Folder"

if os.path.isfile(path):
    print("The current path belongs to a file")
elif os.path.isdir(path):
    print("The current path belongs to a directory")
else: 
    print("The current path is neighther a file or a folder")
```

```python
## The complete path is going to be same as the abosolute path - os.path.abspath()

## os.path.abspath() - It provide us with the complete path of the absolute pathof the file or directory

file_name='Total_data.txt'  #Total_data.txt -- Relative Path
complete_path = os.path.abspath(file_name)

print("The complete path or absolute path for the given file is:", complete_path)
```
