# Modules

## A module is a single Python file (.py) containing definitions (functions, classes, variables) and statements that you can reuse in other files.

## Types of modules

   1. Built-in: math, os, sys, random, datetime, json
   2. Third-party: installed with pip, e.g. numpy, requests
   3. User-defined: the ones you write yourself
