## Key concepts

   1. try: code that might fail --> try block is going to have the operation where we can expect an error to occur

   2. except: handles a specific error --> which is used to handle the error in a way such that the program execution shouldn't stop

   3. else: runs only if no exception occurred --> it will be consisting of the code which we want to run if the try block operation is working without throwing any error
   
   4. finally: always runs (cleanup) --> This blocks gets executed once we are done with everything and even if we are getting an error or not getting an error still the code inside, finally block gets executed. 

```python
# example - 

num = int(input("Enter the denominator value"))
try:
    result=1/num
except Exception as ex:
    print(ex)
else:
    print("The division result for your denominator value will be ", result)
```
