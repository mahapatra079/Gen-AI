# Dictionary Methods

## Dictionary Methods

1. keys()
2. values()
3. items()
4. get()

```python
# keys - method returns a view object. The view object contains the keys of the dictionary, as a list.

## When an item is added in the dictionary, the view object also gets updated:

# Keys - This is a method in dictionary which returns all of the keys present inside our dictionary.

# Example 

car = {
  "brand": "Ford",
  "model": "Mustang",
  "year": 1964
}

x = car.keys()

car["color"] = "white"

print(x)
```

```python
# Value() - it will return all values inside our dictionary

# The values() method returns a view object. The view object contains the values of the dictionary, as a list.

# example

car = {
  "brand": "Ford",
  "model": "Mustang",
  "year": 1964
}

x = car.values()

car["year"] = 2018

print(x)
```

```python
# items() - it will be used for retiurning the key value pairs together.

# The items() method returns a view object. The view object contains the key-value pairs of the dictionary, as tuples in a list.

# example 

car = {
  "brand": "Ford",
  "model": "Mustang",
  "year": 1964
}

x = car.items()

car["year"] = 2018

print(x)
```

```python
# get() - It will be used for returning the value for the key we are going to pass inside our get method.print

#  The get() method returns the value of the item with the specified key.

# Syntax - dictionary.get(keyname, value)

# Example 

car = {
  "brand": "Ford",
  "model": "Mustang",
  "year": 1964
}

# x = car.get("price", 1200)

x = car.get('model')

car["year"] = 2018

print(x)
```

# Shallow Copy - This is being used for creating a copy of the original dictionary in a way such that the changes that we are going to make with the original dictionary won't be getting reflected into our copied dictionary and shallow copy is being done by copy().

# The copy() method returns a copy of the specified dictionary.

```python
# Copy() Syntax - dictionary.copy()

# example

car = {
  "brand": "Ford",
  "model": "Mustang",
  "year": 1964
}

x = car.copy()

print(x)
```

# Nested Dictionary 

```python
# Nested Dictionary will be used for keeping a dictionary inside another dictionary.

## I want to store the information for multiple students like their name, marks, grade

# example 1 -  # Object of objects

Nested_dictionary = {
    "Student_1": {"Name": "Amit", "Marks": 84, "Grade": "A"},
    "Student_2": {"Name": "Vivek", "Marks": 93, "Grade": "A+"},
    "Student_3": {"Name": "Rahul", "Marks": 75, "Grade": "B"}
}

print(Nested_dictionary["Student_2"]["Name"])

for key, student in Nested_dictionary.items():
    print(key, student["Name"], student["Marks"], student["Grade"])
```

```python
# Example 2 - # Array of objects

students = [
    {"name": "Aarav", "marks": 92, "grade": "A"},
    {"name": "Priya", "marks": 85, "grade": "B"},
    {"name": "Rohan", "marks": 67, "grade": "D"},
]

for s in students:
    print(s["name"], s["marks"], s["grade"])
```

# Dictionary Comprehension

```python
# Dictionary Comprehension helps us with using Looping statement along with Conditional statement inside our dictionary.

# Syntax - {key: value for item in iterable}

## Example

## Squares of numbers
squares = {x: x * x for x in range(1, 6)}
print(squares)

## With a condition (only even numbers):
even_squares = {x: x * x for x in range(1, 11) if x % 2 == 0}
print(even_squares)
```

# Merged Dictionary

```python
# example 

dict1 = {"Amit": 84, "Vivek": 93}
dict2 = {"Rahul": 75, "Sneha": 88}

merged = {**dict1, **dict2}  ## unpacking method 
print(merged)
```

```python
# Create a dictionary price_dict where keys are product names and values are prices

price_dict = {
  "Mouse": 240,
  "charger": 1320,
  "mobile": 35000,
  "laptop":85000,
  "keyboard":520,
  "usb-cable":600
}
print(price_dict)
```

```python
# Write a small code blocks with price_dict.
   # Add a new product with price to pricd_dict.
      #  - Update the price of an existing product
      #  - Remove a product by name (handle the case when the product does not exist)

# Add a product
price_dict["Monitor"] = 12000

# Update an existing product
price_dict["Mouse"] = 260

# Remove a product safely
product_to_remove = "charger"

if product_to_remove in price_dict:
    del price_dict[product_to_remove]
    print(f"{product_to_remove} was removed.")
else:
    print(f"{product_to_remove} was not found.")

print(price_dict)
```

```python
# Print the average price of all products (use only dictionary operations and basic arithematic)
   # extra(optional): Print the product with both the maximum and minmum prices.

products = {
    "laptop": 55000,
    "phone": 22000,
    "headphones": 1500,
    "keyboard": 800,
    "monitor": 12000,
}

# Average price: sum of values divided by count
total = 0
for price in products.values():
    total += price

average = total / len(products)
print("Average price:", average)

# Extra: products with the maximum and minimum prices
max_price = max(products.values())
min_price = min(products.values())

for name, price in products.items():
    if price == max_price:
        print("Most expensive:", name, price)
    if price == min_price:
        print("Cheapest:", name, price)
```

```python
# combined operation

products = list(price_dict.keys())

category_by_product = {
    "Mouse": "Accessories",
    "charger": "Electronics",
    "mobile": "Electronics",
    "laptop": "Electronics",
    "keyboard": "Accessories",
    "usb-cable": "Accessories",
    "Monitor": "Electronics",
}

catalog = []

for product in products:
    catalog.append((product, price_dict[product], category_by_product[product]))

print(catalog)
```

```python
# From catalog, create a new dictionary category_to_products 
  # that maps each category to a list of product names in that category.

category_to_products = {}

for product_name, price, category in catalog:
    if category not in category_to_products:
        category_to_products[category] = []

    category_to_products[category].append(product_name)

print(category_to_products)
```

```python
# Print all the products that belong to the category that has the maximum number of products

max_count = 0

for products in category_to_products.values():
    if len(products) > max_count:
        max_count = len(products)

for category, products in category_to_products.items():
    if len(products) == max_count:
        print(f"{category} ({max_count} products): {products}")
```
