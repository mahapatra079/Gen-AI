# CSV Library

## csv library  - Most of the time we are going to work with CSV files so thats why knowledge with csv library functionalties matters.

## The csv module is built into Python, so no installation is needed.

## import csv

```python
# example - csv library 

import csv

with open("example.txt",mode ='w', newline='') as file:
    writer = csv.writer(file)
    writer.writerow(['name','age'])
    writer.writerow(['Amit',23])

with open('example.txt',mode = 'r') as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)
```

## JSON Library 

## When we are going to work with NOsql Database data then the data inside the Nosql Database is going to be their in the JSON format and along with that when we are going to work with the datasets as well then again these dataset will be consisting of data in the JSON format itself.

## When we get the JSON DATA, we get in a JSON string format so by using JSON library we can convert our string data into dictionary format for serializing our data. 

```python
# example - converting from Sting format data to Dictionary format data

import json

data = {"Name": "Amit","Age":31}
json_str = json.dumps(data)
print(json_str)
print(type(json_str))
```

```python
# load

json_dict = json.loads(json_str)
print(json_dict)
print(type(json_dict))
```
