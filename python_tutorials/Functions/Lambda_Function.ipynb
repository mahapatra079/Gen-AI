# Lambda functions

# Lambda Functions - can take any number of arguments, but can only have one expression.

# defined in a single line using the lambda keyword.

## Syntax 

lambda arguments : expression

# The expression is executed and the result is returned:


```python
# example - Add 10 to argument a, and return the result:

x = lambda a : a + 10
print(x(5))

# Lambda functions can take any number of arguments
```

```python
# Example - Multiple arguments
# adding two numbers

add = lambda a, b: a + b
print(add(3, 4))
```

```python
# example  - No arguments

greet = lambda: "Hello!"
print(greet()) 
```

```python
# Example - Default arguments
power = lambda x, n=2: x ** n
print(power(3))    
print(power(3, 3)) 
```

# Lambdas are best for short, simple, one-off functions, especially as arguments to higher-order functions like sorted, map, and filter. For anything more complex or reusable, use a regular def function.

# Lambda with Built-in Functions - Lambda functions are commonly used with built-in functions like map(), filter(), and sorted().
