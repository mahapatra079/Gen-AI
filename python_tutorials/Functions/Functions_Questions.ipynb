```python
# Example 1 - To create a function that converts  temperature from celsius to Fahrenheit or
# Fahrenheit to celsius
# Formulas: F = C × 9/5 + 32 and C = (F − 32) × 5/9.

# Simple function

def celsius_to_fahrenheit(c):
    return c * 9 / 5 + 32

def fahrenheit_to_celsius(f):
    return (f - 32) * 5 / 9

print(celsius_to_fahrenheit(100))   
print(fahrenheit_to_celsius(98.6)) 
```

```python
# example - a single function with a direction parameter fits well:

def convert_temperature(value, unit):
    """Convert a temperature.

    value: the number to convert
    unit:  "C" if value is in Celsius (returns Fahrenheit),
           "F" if value is in Fahrenheit (returns Celsius)
    """
    unit = unit.strip().upper()

    if unit == "C":
        return value * 9 / 5 + 32
    elif unit == "F":
        return (value - 32) * 5 / 9
    else:
        raise ValueError("unit must be 'C' or 'F'")

print(convert_temperature(100, "C")) 
print(convert_temperature(212, "F")) 
print(convert_temperature(37, "c"))
```

```python
# Example 2 - Define a Function that would behave as a password strength checker.
## Password Strength Checker

'''
 Strong Password--->
 if your password:
    1. length is greater than 8 characters 
    2. password must consist of atleast a lowercase (a,b,c...z)
    3. password must consist of atleast a uppercase (A,B,C,..Z)
    4. password must consist of atleast a digit (1,2,)
    5. password must consist of atleast a special character (!,@,#,$,%,^,&,*,(  ),_)
    
 Weak Password--->
 if a password doesn't match any above condition then it will be considered as a weak password 
 '''
def password_strength(password):

    special_chars = "!@#$^&*()_"

    if len(password) < 8:
        print(f"The entered password {password} is a Weak Password")

    elif not any(char.islower() for char in password):
        print(f"The entered password {password} is a Weak Password")

    elif not any(char.isupper() for char in password):
        print(f"The entered password {password} is a Weak Password")

    elif not any(char.isdigit() for char in password):
        print(f"The entered password {password} is a Weak Password")

    elif not any(char in special_chars for char in password):
        print(f"The entered password {password} is a Weak Password")

    else:
        print(f"The entered password {password} is a Strong Password")


password = input("Enter a password: ")
password_strength(password)
```

# Palindrom String

## A palindrome string is going to have characters similars to from both the sides or even if you will reverse the string still you will be getting the given string itself.

## palindrome--> aba, aaabbaaa, zzyzz
## not a palindrome---> Ab, abaa

```python
# example 3 - check that whether if the current string is a plaindromeor not
# Palindrome Checker

def palindrome_check(string):

    if string == string[::-1]:
        print(f"The entered string {string} is a Palindrome")
    else:
        print(f"The entered string {string} is not a Palindrome")


string = input("Enter a string: ")
palindrome_check(string)
```

```python
# example 4 - Calculate the total cost of items in a Shopping Cart

## index in our cart list ---> keyname
## Total ---> Cost of 1 item *Quantity

cart = [{'item1': 'Apple',  'Cost': 5, 'Quantity': 20},
        {'item2': 'Banana', 'Cost': 4, 'Quantity': 15},
        {'item3': 'Guava',  'Cost': 3, 'Quantity': 10} ]

def cart_total(cart):
    total = 0

    for item in cart:
        total += item['Cost'] * item['Quantity']

    return total
print(f"The total cost of items in the shopping cart is {cart_total(cart)}")    
```
