# Map Function

# Lambda functions are commonly used with built-in functions like map(), filter(), and sorted().

# map() - The map() function applies a function to every item in an iterable:

# Map Function applies a given function operation to All items in an iterable or a list and returns a MAP OBJECT (iterable)

# Syntax - 
  # list (map(function,iterable))
  # map(function, iterable, *more_iterables)

```python
# Example - Squaring all of the item list 

def square(list_items):
    empty_list = []
    for item in list_items:
        squared_item = item**2
        empty_list.append(squared_item)
    print(empty_list)
square([1,2,3,4,5,6,7,8,9])
```

```python
# Example - Squaring all of the item list with Lambda Function in built MAP Function

nums = [1, 4, 5, 3]
print(list(map(lambda x: x * 2, nums)))
```

## While mapping the value of one iterable with another iterable we can actually apply the operation in between the values of the two iterables in a synchronous way.

[1,2.3.4]<--->[5,6,7,8]---> Addition ---> [6,8,11,6]

```python
# example - Map function  with Multiple Iterables

num_1 = [1,4,5,8]
num_2 = [8,5,6,7]

list(map(lambda x,y:x+y,num_1,num_2 ))
```

```python
# example - Map() to convert all string values items in an iteratable into an integer

list(map(int,['1','2','3','4','5','6','7','8']))
```

```python
# example - converting the list into Uppercae

user_list = ['amit','rahul','rohit','sumit','rohan']
list(map(str.upper,user_list))
```

```python
list(map(str.lower,user_list))
```

```python
# example - Retrieve the values inside a dictionary using map()

people = [
    {"name": "Alice", "age": 25},
    {"name": "Bob", "age": 30},
    {"name": "Cara", "age": 22},
]

list(map(lambda p: p["name"], people))
list(map(lambda p: p["age"], people))
```
