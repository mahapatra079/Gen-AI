# Streamlit in Python


## Streamlit is an open-source Python framework used to build interactive web applications quickly, especially for Data Science, Machine Learning, and Generative AI projects.

## The best part is that you can build a web application using only Python without needing HTML, CSS, JavaScript, or React.

## streamlit run app.py - It will automatically open your application in the browser.

# Why to use Streamlit? 
   
   1. Easy to use
   2. fast development
   3. Interactive UI Elements  ---> Buttons, Charts, etc
   4. Great for data structure  ---> supports pandas,matplotlib, plotly and other libraries 
   5. Deployment is easy

## Install Streamlit - https://streamlit.io/

# First, install Streamlit using pip: pip install streamlit

# Check the installed version: streamlit --version

# Your first Streamlit application - Create a Python file named app.py.

# Run the application: -  streamlit run app.py

# Important Streamlit functions

    Function	              Purpose

    st.title()	             Display a title
    st.header() 	         Display a heading
    st.write()	             Display text, data, etc.
    st.text_input()	         Accept text input
    st.number_input()	     Accept numbers
    st.button()	             Create a button
    st.selectbox()	         Create a dropdown
    st.checkbox()	         Create a checkbox
    st.slider()	             Create a slider
    st.image()	             Display images
    st.dataframe()	         Display tabular data
    st.sidebar	             Create a sidebar
    st.success()	         Display success message
    st.error()	             Display error message

### Note: Before running the app, make sure app.py is in the Streamlit folder. Run the command from that folder, or provide the correct path to app.py. Otherwise, Streamlit will report that it can’t find the file.
