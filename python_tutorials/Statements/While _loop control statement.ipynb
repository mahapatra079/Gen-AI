## Looping Statements While loop and loop control statements


# While Loop

# While Loop - With the while loop we can execute a set of statements as long as a condition is true.

# Syntax  - # While Condition: 
#                     # code to repeat

```python
# While Loop Example

count = 1

while count <= 5:
    print(count)
    count += 1
```

```python
# Infinite Loop

i = 1

while i > 0:
    print(i)
    i +=1
```

# Loop Control Statements 

# Loop Control Statements are being used for terminating a loop.

# Loop control statements change the normal flow of a loop. 

# Python has three: break, continue, and pass.

```python
# Break - Exits the loop immediately, even if the loop condition is still true.

# example 

for i in range(1, 10):
    if i == 5:
        break
    print(i)
```

```python
# Print the numbers from 0 to 50 but once number = 18, then it will terminate the loop or do not print
# either 18 or any number forward

for i in range(51):
    if i == 18:
        break
    else:
        print(i)
```

```python
# Continue - Skips the rest of the current iteration and jumps to the next one. The loop itself keeps going.

# Example

for i in range(1, 6):
    if i == 3:
        continue
    print(i) 
```

```python
# example - Print only odd numbers:

for i in range(1, 10):
    if i % 2 == 0:
        continue
    print(i)   
```

```python
# example - print the numbers from 0 to 50 but only print even numbers and for odd numbers skip those values 

for i in range(51):
    if i % 2 != 0:  # odd number -> skip it
        continue 
    print(i)
```

```python
# print the numbers from 0 to 20 but once 17 is occuring skip it

for i in range(21):
    if i == 17:
        continue
    else:
        print(i)
```

```python
# pass - Does nothing. It’s a placeholder where Python needs a statement but you have no code yet.

# Example - print the numbers from 0 to 20 but once 17 is occuring do not

for i in range(21):
    if i == 17:
        print("will not do nothing also it won't print")
        pass
    else:
        print(i)
```

```python
# example - Division of numbers where the denomintor is 0

num_1 = int(input("Enter the numerator value: "))
num_2 = int(input("Enter the denominator value: "))

if num_2==0:
    print("do nothing")
    pass
else:
    print(num_1 / num_2)
```

# Nested Loop

# Nested Loop - Its going to have a loop inside a loop

## A nested loop is a loop inside another loop.
# For each single iteration of the outer loop, the inner loop runs completely from start to finish.

## fstring - This will be helping us out with printing the numbers with the variable value present inside the colons so we can
# directly represent the variable value inside the colons mesaage itself where variable value will be shown inside the braces{}

```python
# example - Nested loop can help us out  with iterating over a series of numbers as with the proper combination of iterating numbers.

for i in range(6):
    for j in range(4):
        print(f"i value is: {i} and j value: {j}")
```

```python
# exmaple - print the sum of first N natural number where ---> 1,2,3,......,n
# range(1, n + 1) is used because the stop value is not included.

n = int(input("Enter the N value: "))

sum = 0

for i in range(1,n+1):
    sum +=i
print(f"the sum of first {n} natural number is: {sum}")
```

```python
# print the prime numbers from 0 to 100

for num in range(2, 101):
    for i in range(2, num):
        if num % i == 0:
            break               # found a divisor, not prime
    else:
        print(num, end=" ")     # no break happened, so it's prime
```
