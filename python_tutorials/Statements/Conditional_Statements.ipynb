# Conditional Statements


# Conditional Statements  -  It works like Real Life scenario where if a certain condition is meeting them a statement of code or a block of code gets executed.

# Conditional statements let your program make decisions: run certain code only when a condition is true.

# Conditional Statement

1. if statement
2. elif statement
3. else

```python
# example 1 - with if condition

age = 20

if age >= 18:
    print("Adult")
```

```python
# example 2 - with if else condition

age = 15

if age >= 18:
    print("Adult")
else:
    print("Minor")
```

```python
# example 3 if / elif / else - to check multiple conditions in order. The first true one runs, and the rest are skipped.

marks = 72

if marks >= 90:
    grade = "A"
elif marks >= 75:
    grade = "B"
elif marks >= 60:
    grade = "C"
else:
    grade = "F"

print(grade)  
```

```python
# example  4 - Figure out who is the tallest among three friends

amit_height = int(input("Enter the height of Amit:"))
rahul_height = int(input("Enter the height of Rahul:"))
# rohit_height = int(input("Enter the height of Rohit:"))

if amit_height > rahul_height:
    print("Amit is taller than Rahul")
else:
    print("Rohit is taller than Amit")
```

```python
# example 5 - Figure out who is the tallest among three friends using multiple condition

amit_height = int(input("Enter the height of Amit:"))
rahul_height = int(input("Enter the height of Rahul:"))
rohit_height = int(input("Enter the height of Rohit:"))

if amit_height > rahul_height > rohit_height:
    print("Amit is tallest among the three friends")

elif  rahul_height > amit_height > rohit_height:
    print("Rahul is the tallest among the three friends")

else:
    print("Rohit is the tallest among the three friends")
```

# Nested Conditional Statements 

```python
# Nested Conditional Statements - it involve if statements inside another if statements.

# example 6

amit_height = int(input("Enter the height of Amit:"))
rahul_height = int(input("Enter the height of Rahul:"))
rohit_height = int(input("Enter the height of Rohit:"))

if amit_height > rahul_height:
    if rahul_height > rohit_height or amit_height > rohit_height:
     print("Amit is tallest guy")

elif  rahul_height > amit_height:
      if rahul_height > rohit_height or amit_height > rohit_height:
       print("Rahul is the tallest among the three friends")

else:
    print("Rohit is the tallest among the three friends")
```

```python
# Whether the current year is leap year or not 

year = int(input("Enter the year value: "))

if year % 4 == 0:
    if year % 100 == 0:
        if year % 400 == 0:
            print(year, "is a leap year")
        else:
            print(year, "is not a leap year")
    else:
        print(year, "is a leap year")
else:
    print(year, "is not a leap year")
```

```python
# without Nesting  - # Whether the current year is leap year or not 

year = int(input("Enter the year value: "))

if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
    print(year, "is a leap year")
else:
    print(year, "is not a leap year")
```
