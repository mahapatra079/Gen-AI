# Loop statements are used for iterating over a series of numbers or through iterables.


```python
# example - print numbers from 1 to 5

print(1)
print(2)
print(3)
print(4)
print(5)
```

# range()

# range() -  range function can be used for working as our iterable

# Generates a sequence of numbers.

# range() generates a sequence of numbers on demand. It has three forms:

# Syntax  - range(stop)                 # 0 up to stop-1
#         - range(start, stop)          # start up to stop-1
#         - range(start, stop, step)    # start up to stop-1, jumping by step



```python
# range() example 1 - 

for i in range(5):
    print(i)   
```

```python
# range() example 2

for i in range(1, 6):
    print(i)    
```

```python
# range example 3

for i in range(0, 10, 2):
    print(i) 
```

```python
# range - Counting backwards

for i in range(5, 0, -1):
    print(i) 
```

```python
for  i in range (0,100,5):
    print(i)
```

```python
# Looping by index

fruits = ["apple", "banana", "cherry"]

for i in range(len(fruits)):
    print(i, fruits[i])
```

# Looping statements, we can iterate through iterable like String as well
