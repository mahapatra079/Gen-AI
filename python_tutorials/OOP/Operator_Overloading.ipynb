# Operator Overloading


# Operator Overloading is an OOP concept in which we give additional or customized meaning to existing operators when they are used with user-defined objects.

# we can define how operators like +, -, *, ==, and > should work with our own classes.

# Operator overloading is implemented using Magic Methods (Dunder Methods) in Python.

# Why do we need Operator Overloading?

   1. Makes custom objects behave like built in types(numbers, string, lists).
   2. Improve code readability by using natural numbers syntax(object1 + object2) instead of object1.add(object2)

## operator ----> + ----> Method Name ---> __add__(self,other)
## operator ----> - ----> Method Name ---> __sub__(self,other)
## operator ----> * ----> Method Name ---> __mul__(self,other)
## operator ----> / ----> Method Name ---> __truediv__(self,other)
## operator ----> == ----> Method Name ---> __eq__(self,other)
## operator ----> > ----> Method Name ---> __gt__(self,other)

```python
# example - Overloading the + Operator

class Student:

    def __init__(self, marks):
        self.marks = marks

    def __add__(self, other):
        return self.marks + other.marks


s1 = Student(80)
s2 = Student(90)

print(s1 + s2)

# __add__():
# - self refers to s1.
# - other refers to s2.
# - self.marks = 80.
# - other.marks = 90.
## Therefore, 80 + 90 = 170.
```

```python
# example 

class Avengers:

    def __init__(self,name,power):
        self.name = name
        self.power = power

    def __add__(self,other):
        return Avengers(self.name + " & " + other.name, (self.power + other.power)/2)

    def __sub__(self,other):
        return Avengers(self.name, self.power -other.power)

    def __mul__(self,other):
        return Avengers(self.name, self.power*other.power*2)

    def __truediv__(self, other):
        return Avengers(self.name, self.power/other.power)

    def __str__(self):
        return f"{self.name} with power {self.power}"

thor = Avengers("Thor",100)
hulk = Avengers("Hulk",120)

print(thor + hulk)

## (100 + 120) / 2 - gives 110.0
```

```python
# example - 

class AIModel:
    def __init__(self,name,accuracy):
        self.name = name
        self.accuracy = accuracy

    def __gt__(self, other):
        return self.accuracy > other.accuracy

    def __lt__(self, other):
        return self.accuracy < other.accuracy

    def __eq__(self, other):
        return self.accuracy == other.accuracy

gpt4 = AIModel("GPT4",90)
bard = AIModel("Google Bard", 100)

print(gpt4 < bard)
```
