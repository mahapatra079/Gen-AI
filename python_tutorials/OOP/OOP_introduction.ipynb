# Object Oriented Programming Introduction


# OOP

# OOP stands for Object-Oriented Programming.

# Python is an object-oriented language, allowing you to structure your code using classes and objects for better organization and reusability.

# Advantages of OOP

   1. Provides a clear structure to programs
   2. Makes code easier to maintain, reuse, and debug
   3. Helps keep your code DRY (Don't Repeat Yourself)
   4. Allows you to build reusable applications with less code

## Class and Object  - Classes and objects are the two core concepts in object-oriented programming.

   # Class - A blueprint for creating objects.

   # Objects - An instance of a class. 

```python
# example - 

class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def display(self):
        print(self.name, self.age)

# Creating objects
student1 = Student("Amit", 31)
student2 = Student("Rahul", 25)

student1.display()
student2.display()
```

## Important concepts

   - class: Defines a class.
   - __init__(): Constructor that initializes an object.
   - self: Refers to the current object.
   - self.name: Instance variable.
   - display(): Instance method.

# Object Oriented Programming Concepts


```python
# example

class Cartoon:
    def __init__(self, name, show, catchphrase):
        self.name = name
        self.show = show
        self.catchphrase = catchphrase

    def introduction(self):
        print(f"Hello, my name is {self.name}")

    def speak(self):
        print(f"{self.name} says {self.catchphrase}")

Mickey_mouse = Cartoon("Mickey mouse", "Cartoon show", "Let's have fun!")

Mickey_mouse.introduction()
Mickey_mouse.speak()
```

# Object Oriented Teminologies

   1. Attributes  - This Stands for the characteristics that a specific object is going to have 

   2. Methods  - Functionality that have a specific object is going to have 

   3. self - It refers to the current object of the class 

   4. Constructor - This refers to __init__()

```python
class Car:
    color = "blue"
    tyres = "MRF"
    car_type = "Fully Electric"

car_1 = Car()

print(car_1.color)
print(car_1.tyres)
print(car_1.car_type)
```

# Types of Constructor:

# Types of Method: 

   1. Instance Method - Instance Methods operate on Object Attribute.   
      - this is going to use self keyword to access and modify object attributes
      - These methods are called Using an Object.
   
   2. Class Method - It is the method in which the  whole class or the whole blueprint which will remain the samefor all of the objects then such kind of method is been reffered as Class Method.

    - In class method, these methods works with the class level attributes.
    - they are going to use cls instead of self
    - they are called using class name as well
    - for specific, current method is the class method we have to use the abbreviation as @classmethod

   3. Static Method - Static method do not use self or cls

    - They work like normal functions but belong to a class
    - They are going to be designed by a abbreviation @staticmethod
    - Static method can also be called by classname

```python
# example

class Avengers:

    def __init__(self,name,power):
     self.name=name
     self.power=power

    def show_details(self):
      print(f"My name is {self.name} and I am having power {self.power}")

iron_man = Avengers("Iron man","Genius intellect")

iron_man.show_details()
```

```python
# example

class Avengers:

    origin="Avengers"

    def __init__(self,name,power):
        self.name = name
        self.power = power

    @classmethod
    def origin_info(cls):
        print(f"My origin is {cls.origin}")

    def self_info(self):
        print(f"My name is {self.name} and my power is {self.power}")

Iron_man = Avengers("Iron Man", "Genius Intellect")
Iron_man.power
```

```python
# example  - Static MeThod

class Avengers:

    #static method
    @staticmethod
    def mission():
        print("Avebgers Assembled!...")

    origin  = "Avengers"

    def __init__(self,name,power):
        self.name = name
        self.power = power

    #class Method
    @classmethod
    def origin_info(cls):
        print(f"The origin for Avengers is: {cls.origin}")

    # Instance Method
    def introduction(self):
        print(f"My nameis {self.name} and my power is {self.power}")

iron_man = Avengers("Iron Man","Genius Intelect")

iron_man.introduction()
iron_man.origin_info()
iron_man.mission()
```
