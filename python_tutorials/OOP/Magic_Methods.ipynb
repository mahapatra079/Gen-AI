# Magic Methods


# Magic Methods  - Dunder Methods

## Magic methods are special built-in methods in Python that allow objects to interact with Python's operators, functions, and language features.

## They are also called Dunder Methods because their names begin and end with double underscores (__).

## Examples:
   - __init__()  - object intialization
   - __str__()   - String Representation
   - __len__()   - Length
   - __add__()   - Arithematic Operation
   - __repr__()  - String Representation

## Magic methods allow us to define how our own objects should behave when we use Python operations.

```python
# example  - __init__() and __str__() are both magic methods:

class Person:
  def __init__(self, name):
    self.name = name

  def __str__(self):
    return f"Person: {self.name}"

p1 = Person("Email")
print(p1)
```

## Why do we use Magic Methods?

   1. make objects behave like built in types (ex:list, integers)
   2. Improves Readability by allowing basic operations object1 + object2
   3. Simplifies code by replacing verbose function calls with natural syntax

## print(object)  -->  str --> what is returned when print(object) is used

## Arithematic operation can be done among the objects by using Maigc Methods.

```python
## __add__

class Avengers:

    def __init__(self,name,strength):
        self.name = name
        self.strength = strength

    def __add__(self,other):        
        return Avengers(self.name + " " "&" " " + other.name, self.strength + other.strength)

    def __str__(self):        
        return  f"{self.name} with combined strength, {self.strength}"

ironman = Avengers("Ironman",80)
thor = Avengers("thor",100)

team = ironman + thor
print(team)
```
