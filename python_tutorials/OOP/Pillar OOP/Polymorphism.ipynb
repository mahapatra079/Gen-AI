# Polymorphism

## polymorphism allows the same method, function, or operator to behave differently depending on the object or data type.

## polymorphism allows object of different classes as if they are the objects of the same class & it enables reusability and makes programs more flexible.

## Type of polymorphism : 
   
   1. Overiding             - (Runtime Polymorphism)
   2. Overloading           - (Compile Time Polymorphism)
   3. Operator Overloading

## Method Overriding - A child class provides a specific implementation of a method that has already been defined into the parent class. This allows Dynamic Selection a Runtime.

## Method overloading means having multiple methods with the same name but different parameters.

```python
## example - overriding

# parent class
class Grandparent: 

    def speak(self):
        print("Hello, My nanme is Amit")

# child class
class Parent(Grandparent):

    def speak(self):
        print("Hello, My name is rahul")

class child(Grandparent):

    def speak(self):
        print("Hello, My name is rohit")

grandparent = Grandparent()
parent = Parent()
child_obj = child()

for person in (grandparent, parent, child_obj):
    person.speak()
```

```python
## Overloading - Method overloading means having multiple methods with the same name but different parameters.

class Calculator:

    def add(self, a, b, c=0):
        return a + b + c


obj = Calculator()

print(obj.add(10, 20))
print(obj.add(10, 20, 30))
```
