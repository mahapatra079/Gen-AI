# Abstraction 

## Abstraction

# Abstraction means hiding the internal implementation details and showing only the essential functionality to the user.

# This helps in reducing programming commplexity and effoerts.

# Abstraction allow us to work with an object with what an object does rather than how it does it

## How to achieve abstraction in Python?

   1. Abstract Class    - A class that cannot be instantiated directly when it contains abstract methods.
   2. Abstract Method   - A method declared in the parent class without providing its actual implementation. Child classes must implement it.


## The abstract method is being defined by @abstractmethod.

```python
# example - Abstraction

from abc import ABC, abstractmethod

# Abstract class
class Animal(ABC):

    @abstractmethod
    def sound(self):
        pass


# Child class
class Dog(Animal):

    def sound(self):
        print("Dog barks")


class Cat(Animal):

    def sound(self):
        print("Cat meows")


dog = Dog()
cat = Cat()

dog.sound()
cat.sound()
```
