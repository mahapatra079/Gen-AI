# Encapsulation 

# Encapsulation

## Encapsulation means wrapping data (variables) and methods (functions) together inside a class and restricting direct access to some of that data.

## It protects data from being accessed or modified directly and provides controlled access through methods.

## Access modifiers in Python

   1. Public             -  name     - Accessible from anywhere
   2. Protected          -  _name    - Intended for use within the class and subclasses
   3. Private            -  __name   - Name-mangled to discourage direct access
   4. Getter & Setter 

# Public Method

## Public Method - Public variables and methods can be accessed from anywhere
# It can be accessed outside the class.

```python
# example - Public Method

## Here, name is public, so it can be accessed outside the class.

class Student:
    def __init__(self):
        self.name = "Amit"

    def display(self):
        print(self.name)


s = Student()

print(s.name)
s.display()
```

```python
# example - public method

class IronMan: 

    def __init__(self,name,power,weapon):
        self.name = name
        self.power = power
        self.weapon = weapon

    def introduction(self):
        print(f"Hello my name is {self.name} and I have {self.weapon}")

iron = IronMan("Amit", "Genius intellect", "Repulsor")
iron.introduction()
```

```python
dir(iron)
```

# Private Method

## Private Method - Private members use double underscores __.

## Python applies name mangling to make accidental access more difficult.

```python
# example - Private Method

class BankAccount:
    def __init__(self):
        self.__balance = 10000

    def get_balance(self):
        return self.__balance


account = BankAccount()

print(account.get_balance())
```

```python
print(account.__balance)
```

## Because Python internally changes the attribute name to _BankAccount__balance.

## This is called Name Mangling.

## Note: Name mangling is not absolute security; the attribute can still be accessed using its mangled name.

## A private member is not accessibile outside the class and not even in the derived class and in order to have the access over these private members or to bring some modifcation on these memebers we use getter and setter method.

## Private Members can be decalred by double underscore __before the name.

```python
## example  - Private

class Avenger :

    def __init__(self,name,phone_number):
        self.name = name
        self.phone_number = phone_number

avenge=Avenger("captain",'8074942363')
avenge.name
```

```python
dir(avenge)
```

# Getter Method

## getter Function  - it will be helping us out with accessing the private

```python
# example - getter method

class Captain:
    def __init__(self,name,address):
        self.name = name
        self.address = address

    ## getter method

    def accessing_add(self):
        print(f"My address is {self.__address}")

captain = Captain("Captain America", "New York")
captain.address
```

```python
dir(captain)
```

# Setter Method

## Setter Method -> This is being used for modifying our private data.

```python
# example  - setter method 

class Animal:

    def __init__(self,name,sound):
        self.name = name
        self.sound = sound

    def sound_info(self): 
        print(f"The animalis goingto sound as: {self.sound}")

    def sound_modify(self):
        print(f"New Sound for the Animal is : {self.__sound}")

animal=Animal("Bruno","woof woof")

animal.name
```

```python
animal.sound
```

# Protected Data

## Protected data - Protected members are indicated by a single underscore _.

# They are intended to be accessed within the class and its subclasses.

```python
# example - Protected Method

class Parent:
    def __init__(self):
        self._salary = 50000


class Child(Parent):
    def display(self):
        print(self._salary)


obj = Child()
obj.display()

## _salary is a protected member by convention.
## The child class can access it.
## Python does not strictly prevent external access.
```

## Protected Data is the data which is not accessible outside the class. But the protected data is going to be accessible by the derived class or child class

## For having the protected memeber you can mention your protected member as _attribute or by using a single underscore.

```python
## example - Protected Methods

class GrandParent:

    def __init__(self,name,locker_key):
        self.name = name
        self.locker_key = locker_key

    def locker_info(self):
        print(f"The locker key info is {self._locker_key}")
```

```python
class Parent(GrandParent):
    def __init__(self,name,locker_key):
        super().__init__(name,locker_key)

    def child_reveals(self):
        print(f"The locker key info is {self._locker_key}")
```

```python
grandpa = GrandParent("Amit","121315")
```

```python
grandpa.name
```
