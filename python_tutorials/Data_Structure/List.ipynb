## List 

# Lists are ordered, mutable, and can contain elements of different types.

# used to store multiple elements.

# Lists are created using square brackets []:

```python
## List Indexing - returns the position at the first occurrence of the specified value.

#example

lists = [1, 2, 3, 4, 5]
print(lists[0])
print(lists[1])
print(lists[2])
print(lists[3])

## By using the proper indexing value we can retrieve the data from our list.
# The indexing value starts from 0 and goes up to n-1 where n is the total number of items in the list.
```

```python
# length - length of a data structure is going to be the total items we are having in that data structure.
# In this case, we are having 5 items in the list.
```

```python
# Negative Indexing - returns the position at the last occurrence of the specified value.

# example

lists = [1, 2, 3, 4, 5]
print(lists[-1])
```

```python
# For having the access for all of the values you are going to store inside the list, you can use  :

# example

print(lists[:]) # it will return all of the values inside the list.
```

# List Slicing

```python
# List Slicing - returns a slice of the list.

# example

list = [1, 2, 3, 4, 5]
print(list[1:4]) # it will return the values from index 1 to index 3. (4 is excluded)
```

```python
# Reverse Slicing - returns a slice of the list in reverse order.

# example
list = [1, 2, 3, 4, 5]
print(list[::-1]) # it will return the values from index 4 to index 0
```

# List Mutuability

```python
# List Mutuability - lists are mutable, meaning that we can change their content without changing their identity.

# example

list = ['apple', 'banana', 'cherry']
list[1] = 'kiwi'
print(list) # it will return ['apple', 'kiwi', 'cherry']
```

# List Methods


```python
# append() - adds an element at the end of the list.

list = ['apple', 'banana', 'cherry']
list.append('orange')
print(list) # it will return ['apple', 'banana', 'cherry', 'orange
```

```python
# insert() - adds an element at the specified position.

list = ['apple', 'banana', 'cherry']
list.insert(1, 'orange')
print(list) # it will return ['apple', 'orange', 'banana', 'cherry']
```

```python
# remove() - removes the first item with the specified value.

list = ['apple', 'banana', 'cherry']
list.remove('banana')
print(list) # it will return ['apple', 'cherry']
```

```python
# pop() - removes the element at the specified position.

# example 

list = ['apple', 'banana', 'cherry']
list.pop(1)
print(list) # it will return ['apple', 'cherry']
```

```python
# index() - returns the index of the first element with the specified value.

list = ['apple', 'banana', 'cherry']
index = list.index('banana')
print(index) # it will return 1
```

```python
# count() - returns the number of elements with the specified value.

list = ['apple', 'banana','banana', 'cherry']
count = list.count('banana')
print(count) # it will return 2
```

```python
# sort() - sorts the list in ascending order.

# example

list = [1, 3, 2, 5, 4]
list.sort()
print(list) # it will return [1, 2, 3, 4, 5]
```

```python
# reverse() - reverses the order of the list.

# example

list = [5, 4, 3, 2, 1]
list.reverse()
print(list) # it will return [1, 2, 3, 4, 5]
```

```python
# clear() - removes all the elements from the list.

# example

list = ['apple', 'banana', 'cherry']
list.clear()
print(list) # it will return []
```

## List Methods

1. append()  -- insert the value at the end of the list
2. insert()  -- insert the value at a specific index location inside our list 
3. remove()  -- remove the value from our list that we are going to pass inside it 
4. pop()     -- remove the value from the list at a specific index location
5. index()   -- it provides the index of the value that we are going to pass inside it if that value is present inside that list
6. sort()    -- its going to help us out with sorting the values in ascending order
7. reverse() -- it can reverse the values present inside our list i.e, value at last index will be at first index now
8. count()   -- it counts the number of occurence of a specific value that we are going to pass inside our list and it returns that count of occurence
9. clear()   -- for removing all of the values that we are storing inside our list.
