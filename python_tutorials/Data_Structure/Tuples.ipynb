# Tuples

# Tuples  - are used to store multiple items in a single variable.

# Tuples are the second Data Structure which can store the data in ordered way just like List but Tuples are not mutuable (unchangeable).

# A tuple is a collection which is ordered and unchangeable.

# When we say that tuples are ordered, it means that the items have a defined order, and that order will not change.

# Tuples are unchangeable, meaning that we cannot change, add or remove items after the tuple has been created.

# Tuple items are ordered, unchangeable, and allow duplicate values.

# Since tuples are indexed, they can have items with the same value.

```python
# Tuples -  Tuples are like list which can store the data inside it in ordered way but this are immutable. 
# It means we can not change the data inside it once it is created. It is defined by using () brackets.

# example 1 
tuple = (1, 2, 3, 4, 5)
print(tuple)

# example 2
tuple2 = ("apple", "banana", "cherry")
# tuple2[1] = "orange" # it will give an error because we can not change the data inside tuple once it is created.
print(tuple2)
```

```python
# Tuples are also capaable of storing the data of any data types.

# example 3
tuple3 = (1, "Amit", 3.14, True) # data types are int, str, float, bool
print(tuple3)
```

```python
# Tuples are going to be ordered or going to be indexed. 
# It means we can access the data inside it by using index number.
# we can access the data  by [1:] or [::-1] or [::-1] or [::2] or [::3] etc. like forward and backward slicing.

## example 4

tuple4 = (10, 20, 30, 40, 50)
# tuple4[2] = 100 # it will give an error because we can not change the data inside tuple once it is created.
print(tuple4[2]) # it will print 30

# example 5
tuple5 = (10, 20, 30, 40, 50)
print(tuple5[::-1]) # it will print (50, 40, 30, 20, 10)

# example 6
tuple6 = (10, 20, 30, 40, 50)
print(tuple6[1:3]) # it will print (20, 30)
```

# Tuples Methods 

# Tuple have two methods i.e.

1. count()
2. index()

```python
# count method - it is used to count the number of occurrences of a specific element in a tuple.

# index method - it is used to find the index of the first occurrence of a specific element in a tuple.

# example 7
tuple7 = (1, 2, 3, 4, 5, 1, 2, 1)
print(tuple7.count(1)) # it will print 3 because 1 is present 3 times in the tuple.
print(tuple7.index(2)) # it will print 1 because 2 is present at index 1 in the tuple.
```

# Packing & Unpacking Tuples

```python
# Packing & Unpacking Tuples

# Packing Tuples - When we create a tuple, we normally assign values to it. This is called "packing" a tuple:

# example 8
tuple8 = ("apple", "banana", "cherry") # packing a tuple
print(tuple8)
```

```python
# Unpacking Tuples - But, in Python, we are also allowed to extract the values back into variables. This is called "unpacking"

# example 9 

fruits = ("apple", "banana", "cherry")

(green, yellow, red) = fruits

print(green)
print(yellow)
print(red)
```

```python
# Asterisk * - if the number is less than the number of values, you can add ankii * to the variable name and the values will be assigned to the variable as a list 

# example 10

fruits = ("apple", "banana", "cherry", "strawberry", "raspberry")

(green, *yellow, red) = fruits

print(green)
print(yellow)
print(red)

# If the asterisk(*) is added to another variable name than the last,
# Python will assign values to the variable until the number of values left matches the number of variables left.
```

```python
# Join Tuples - To join two or more tuples you can use the + operator:

# example

tuple1 = ("a", "b" , "c")
tuple2 = (1, 2, 3)

tuple3 = tuple1 + tuple2
print(tuple3)
```

```python
# Multiply Tuples - If you want to multiply the content of a tuple a given number of times, you can use the * operator

# Example

fruits = ("apple", "banana", "cherry")
mytuple = fruits * 2

print(mytuple)
```
