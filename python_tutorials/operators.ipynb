# Operators

# Operators are used to perform operations on variables and values.
# Operators helps in applying operation on the given operand. 


## Types of Operators

1. Arithematic Operator:  used with numeric values to perform common mathematical operations
    
    Addition                     +
    Subtraction                  -
    Multiplication               *
    Division                     /
    Modulus                      %
    Exponentiation               **
    Floor division               //

2. Comparision Operator: used to compare two values

    Equal                         ==
    Not equal                     !=
    Greater than                  >
    Less than                     <
    Greater than or equal to      >=
    Less than or equal to         <=

3. Logical Operato: used to combine conditional statement
    
    and             Returns True if both statements are true
    or              Returns True if one of the statements is true
    not             Reverse the result, returns False if the result is true

```python
#Arithematic Operator Example
a=10
b=2
print(a + b)
```

```python
# Modulus Operator

## Modulus operator will be helping us to get the remainder of the division of two numbers.
## left after dividing the first value with second value. It is represented by the symbol %.

a = 10
b = 5
print(a % b)
```

```python
# exponential operator

a = 10
b = 2
print(a ** b)
```

```python
# Equality Operator
## Both values should be equal to each other. 
# If both values are equal then it will return True otherwise it will return False.

# example 1
a = 10
b = 10
print(a == b)

# example 2
a = 10
b = 5
print(a == b)
```

```python
# Not Equal Operator
## Both values should not be equal to each other. 
# It returns True if both values are not equal to each other otherwise it will return False.

#example 1
a = 10
b = 5
print(a != b)

#example 2
a = 10  
b = 10
print(a != b)
```

```python
# Logical Operators
## Logical operators are used to combine conditional statements.

# example 1
a = 10
b = 5
if a > 5 and b < 10:
 print("Both conditions are True", "a is greater than 5 and b is less than 10")

# example 2
a = 10
b = 5
if a > 5 or b < 10:
 print("At least one of the conditions is True", "a is greater than 5 or b is less than 10")
```

```python
# Simple Calculator

a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

operation = input("Enter operation (+, -, *, /): ")

if operation == "+":
    result = a + b
    print("Result:", result)

if operation == "-":
    result = a - b
    print("Result:", result)

if operation == "*":
    result = a * b
    print("Result:", result)

if operation == "/":
    if b != 0:
        result = a / b
        print("Result:", result)
    else:
        print("Error: Division by zero is not allowed.")
```
