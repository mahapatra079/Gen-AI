- Syntax  ---> Using loc[] ---> df.loc[row_labels, column_names]

```python
# Real-World Example: Employee Data Analysis

import pandas as pd

df = pd.DataFrame(
    {
        "Name": ["Amit", "Rahul", "Priya", "John"],
        "Department": ["IT", "HR", "IT", "Finance"],
        "Salary": [50000, 40000, 60000, 45000],
    },
    index=["A", "B", "C", "D"]
)

# 1. Fetch employee names
print(df["Name"])

# 2. Fetch name and salary
print(df[["Name", "Salary"]])

# 3. Fetch first three records
print(df.head(3))

# 4. Fetch employees from IT
print(df[df["Department"] == "IT"])

# 5. Fetch employees earning above 50000
print(df[df["Salary"] > 50000])

# 6. Fetch names and salaries of IT employees
print(df.loc[df["Department"] == "IT", ["Name", "Salary"]])

# 7. Fetch top three salaries
print(df.nlargest(3, "Salary"))

# 8. Fetch employee at row position 2 (the third row)
print(df.iloc[2])

# 9. Fetch a particular employee's salary (row label "C")
print(df.loc["C", "Salary"])
```

## Performing Analysis Through Filtering Data & Adding New Columns and Discussing Other Important Functions


# In Pandas, filtering and adding columns are two of the most important operations in Data Analysis.

# They help us extract meaningful information, create calculated fields, categorize records, and prepare datasets for Machine Learning and GenAI applications.

```python
## Filtering Data

import pandas as pd

data = {
    "Employee_ID": [101, 102, 103, 104, 105, 106, 107, 108],
    "Name": [
        "Amit", "Rahul", "Priya", "John",
        "Sara", "David", "Neha", "Rohit"
    ],
    "Age": [25, 28, 23, 30, 26, 32, 24, 29],
    "Department": [
        "IT", "HR", "IT", "Finance",
        "HR", "IT", "Finance", "Sales"
    ],
    "Salary": [
        50000, 40000, 60000, 45000,
        35000, 65000, 55000, 48000
    ],
    "Experience": [2, 5, 1, 7, 3, 8, 2, 4],
    "City": [
        "Bhubaneswar", "Delhi", "Mumbai", "Pune",
        "Delhi", "Bangalore", "Mumbai", "Bhubaneswar"
    ]
}

df = pd.DataFrame(data)

print(df)
```

```python
# Filtering using a single condition

# Example 1: Employees earning more than ₹50,000.

result = df[df["Salary"] > 50000]

print(result)
```

```python
# Example 2: Employees younger than 25.

result = df[df["Age"] < 25]

print(result)
```

```python
# Example 3: Employees working in IT.

result = df[df["Department"] == "IT"]

print(result)
```

```python
# Example 4: Employees with five years of experience.

result = df[df["Experience"] >= 5]

print(result)
```

```python
# Adding a column using an existing column

# Example: Calculate annual salary if the current Salary represents monthly salary.

df["Annual_Salary"] = df["Salary"] * 12

print(df[["Name", "Salary", "Annual_Salary"]])
```

```python
# Adding a column using mathematical operations

# Example: Calculate a 10% bonus.

df["Bonus"] = df["Salary"] * 0.10

print(df[["Name", "Salary", "Bonus"]])
```

```python
# Example: Calculate salary after adding the bonus.

df["Total_Salary"] = df["Salary"] + df["Bonus"]
```
