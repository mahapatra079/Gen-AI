# Why do we use Pandas?	           

    Feature                   Purpose
    Data Cleaning	    Remove incorrect or duplicate data
    Data Analysis	    Calculate average, maximum, minimum, etc.
    Data Filtering	    Extract specific records
    Data Manipulation	Add, update, and delete columns
    File Handling	    Read CSV, Excel, JSON files
    Data Grouping	    Group records based on categories
    Data Merging	    Combine multiple datasets
