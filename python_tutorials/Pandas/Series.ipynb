# Series

# Series - A Series is a one-dimensional labeled array that can hold different types of data.

```python
# example 

import pandas as pd

data = [10, 20, 30, 40, 50]

s = pd.Series(data)

print(s)
```

# Here:
- Left side represents the index.
- Right side represents the values.
- dtype represents the data type.

# Pandas Series is a Column of data in a table like Excel or Google Sheet with labels called as indexes attached to each value.

```python
# example - 

import pandas as pd

pd.Series(data=[40, 50, 60], index=['Bread', 'Milk', 'Butter'], name="List of items and price")
```

## Series is a 1D Array that can hold any type of data like integer, float, strings (objects), boolean, mixed data types.

## Data Types 
  
   - Integer --> Data in a Series (column) ---> Quantity of items in the stock
   - FLoat   --> Data in a Series (Column) ---> Price or Product Ratings
   - String  --> column  ---> Product Name
   - Boolean --> Column  ---> Is item in stock?
   - Date / time --> Column ---> Expiry Dates or Sale Date
   - Mixed Types --> Milk, 40, True

# Dictionary to Series

# Dictionary is another Data Structure that stores the data in a key value pair and just because it stores the data in key value pair so we dont have to explicity mention in the index.

```python
# Dictionary to Series

data = {
    "John": 85,
    "David": 90,
    "Alex": 75,
    "Sam": 95
}

s = pd.Series(data)

print(s)
```

# Explanation:

  - Dictionary keys become Series indexes.
  - Dictionary values become Series values.
  - Indexes are automatically assigned from dictionary keys.

```python
# Creating a Series Using a Custom Index

# We can manually specify index labels using the index parameter.

data = [85, 90, 75, 95]

s = pd.Series(
    data,
    index=["John", "David", "Alex", "Sam"]
)

print(s)
```

# Explanation:

- index assigns custom labels.
- The number of index labels must match the number of values.
- We can access values using custom labels.

```python
print(s["John"])
```

```python
# Creating a Series Using a Scalar Value

# We can create a Series using a single scalar value.

# However, we must provide an index to determine how many times the scalar value should be repeated.

s = pd.Series(10, index=[0, 1, 2, 3, 4])

print(s)
```

# Explanation:

- 10 is the scalar value.
- index determines the number of elements.
- Pandas repeats the scalar value for every index.

```python
# Example 

s = pd.Series(
    100,
    index=["A", "B", "C"]
)

print(s)
```

    Method	    Description
    
    head()	    Returns first 5 elements

# Mathematical Functions in Series and indexing in Series


## Mathematical Function

   1. sum() --->  sum of all values present in the series 

   2. product()  ---> product of all values in the series    

   3. mean()   --->  Calculates the average by adding all values and dividing by the number of values (average value)
   
   4. median() --> Returns the middle value after sorting the data. If there are an even number of values, it returns the average of the two middle values.

   5. mode() --> Returns the value or values that occur most often. There can be more than one mode.

   6. std() --> Standard Deviation, If i need to get the idea that how the values is getting repeated maximum number of times.

   7. variance --> This also provides the insight about the distribution of the data from the meean value and variance is the square of the standard deviation.

   8. min() --> minimum value in the series
   
   9. max() --> maximum value in the series

   10. describe() --> This function provides the idea about the summary of the important mathematical calculation for our data.

   11. count ---> Total number of rows in the column

```python
# example - Mathematical Functions

s = pd.Series([10, 20, 30, 40, 50], index=['a', 'b', 'c', 'd', 'e'])
```

```python
s.sum()
```

```python
s.mean()
```

```python
s.prod()      # product of all values
```

```python
s.mode()  # most frequent value(s)
```

```python
s.median()
```

```python
s.std()      # standard deviation (sample, ddof=1)
```

```python
s.var()      # variance
```

```python
s.min()   
```

```python
s.max()
```

```python
s.count()    # number of non-null values
```

```python
s.describe()
```
